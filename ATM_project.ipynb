```python
pin = "1234"
balance = 50000

enter_pin = input("Enter pin: ") # input the pin

if enter_pin == pin:    # if entered pin = original then print correct pin else print wrong pin
  print("Correct pin")
  amount = float(input("Enter amount: ")) # enter the amount that to be withdrwan

  if amount <= balance:
    balance = balance - amount          #if withdrwaed amount is < available balance then collet the cash and print the remaining balance
    print("Collect Cash: ", amount)
    print("Remaining Balance:", balance)
  else:
    print("Insufficient Balance")       # else print insufficient balance
else:
  print("wrong pin")
```
